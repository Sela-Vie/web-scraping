```python
from bs4 import BeautifulSoup

# Read the HTML file

followers:BeautifulSoup
following:BeautifulSoup

with open('./followers.html', 'r', encoding='utf-8') as file:
    html_content_followers = file.read()
with open('./following.html', 'r', encoding='utf-8') as file:
    html_content_following = file.read()

followers = BeautifulSoup(html_content_followers, 'html.parser')
following = BeautifulSoup(html_content_following, 'html.parser')
```

```python
account_followers_elements = followers.find_all('span', class_='_ap3a _aaco _aacw _aacx _aad7 _aade')
account_following_elements = following.find_all('span', class_='_ap3a _aaco _aacw _aacx _aad7 _aade')

# print(account_followers_elements)
# print(type(account_followers_elements))
# print(type(account_followers_elements[0]))
```

```python
name_set_followers:set = {x.text for x in account_followers_elements}
name_set_following:set = {x.text for x in account_following_elements}

print(f"FOLLOWERS ({name_set_followers.__len__():07}): {name_set_followers}")
print(f"FOLLOWING ({name_set_following.__len__():07}): {name_set_following}")

print()
only_following:set = name_set_following - name_set_followers
only_followers:set = name_set_followers - name_set_following
print(f"FOLLOWERS ONLY ({only_followers.__len__():07}): {only_followers}")
print(f"FOLLOWING ONLY ({only_following.__len__():07}): {only_following}")
```
